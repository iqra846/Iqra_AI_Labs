```python
print("=== AI TRAVEL AGENT ===")

departure = input("Enter Departure City: ").lower()
destination = input("Enter Destination City: ").lower()
budget = int(input("Enter Your Budget (PKR): "))

if departure == "karachi" and destination == "lahore":
    if budget >= 20000:
        print("Flight Available!")
        print("Fare: Rs. 20,000")
        print("Departure: 10:00 PM")
        print("Arrival: 3:00 AM")
    else:
        print("No flight available within your budget.")

elif departure == "karachi" and destination == "islamabad":
    if budget >= 25000:
        print("Flight Available!")
        print("Fare: Rs. 25,000")
        print("Departure: 7:00 PM")
        print("Arrival: 12:00 AM")
    else:
        print("No flight available within your budget.")

elif departure == "karachi" and destination == "multan":
    if budget >= 19000:
        print("Flight Available!")
        print("Fare: Rs. 19,000")
        print("Departure: 5:30 PM")
        print("Arrival: 8:30 PM")
    else:
        print("No flight available within your budget.")

else:
    print("Sorry, no flight is available for this route.")
```
